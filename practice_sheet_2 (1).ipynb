4. What is the difference between dict.keys() / values() / items() and a list — are they "live" or a
snapshot?

Anwer: dict.keys(), dict.values(), and dict.items() return "view" objects that are live, meaning they

5. Why is set membership testing faster than list membership testing?

Anwer: Set membership testing is faster than list membership testing because sets use hash tables to
look up elements in O(1) constant average time, whereas lists use a sequential scan that scales linearly
with size in O(n) time.

6. What happens in memory when you write b = a where a is a list? Draw or describe it.

Anwer: When you write b = a where a is a list, both b and a point to the same list object in memory.
This means that any modifications made to the list through either variable will affect the same underlying
list. No new list is created; instead, both variables reference the same memory location.

7. What is the difference between a list comprehension and a generator expression — both syntactically and
behaviourally?

Anwer: Syntactically, a list comprehension uses square brackets [] while a generator expression uses
parentheses (). Behaviorally, a list comprehension generates the entire list in memory at once, while
a generator expression produces items one at a time and is more memory efficient for large datasets.

# Rewrite using a dict lookup below (should be O(n + m), not O(n * m)):

```python
orders = [
    {"order_id": 1, "customer_id": 101},
    {"order_id": 2, "customer_id": 102}
]
customers = [
    {"id": 101, "name": "Asha"},
    {"id": 102, "name": "Ravi"}
]
def attach_customer_name(orders, customers):
    customer_lookup = {}
    for customer in customers:
        customer_lookup[customer["id"]] = customer["name"]
    for order in orders:
        order["customer_name"] = customer_lookup.get(order["customer_id"])
    return orders
print(attach_customer_name(orders, customers))
```

# Explanation
The purpose of safe_get() is to safely access a value deep inside nested dictionaries without crashing if any key is missing.


```python
data = {"user": {"profile": {"address": {"city": "Pune"}}}}
def safe_get(d, *keys):
    current = d
    for key in keys:
        if not isinstance(current, dict):
            return None
        current = current.get(key)
        if current is None:
            return None
    return current
print(safe_get(data, "user", "profile", "address", "city"))
print(safe_get(data, "user", "profile", "phone"))
```

 # Beginner-Level Q&A
Q1: What's the difference between append() and extend() on a list?

A: append(x) adds x as a single element (even if x is a list, it's added as one nested item). extend(iterable) adds
each element of the iterable individually.

Q2: Why can't lists be dictionary keys?

A: Dict keys must be hashable, and hashability requires immutability. Lists are mutable, so Python disallows them as keys to
avoid the hash value silently becoming invalid after mutation.

Q3: What is the output of {1, 2, 3} == {3, 2, 1} ?

A: True — sets are unordered, so equality only checks membership, not order.
 Intermediate-Level Q&A

Q4: How would you remove duplicates from a list while preserving order?

A: list(dict.fromkeys(my_list)) — dicts preserve insertion order since 3.7, and fromkeys naturally deduplicates.

Q5: What's the time complexity difference between list.insert(0, x) and deque.appendleft(x) , and why
does it matter?

A: list.insert(0, x) is O(n) because every existing element must shift. deque.appendleft(x) is O(1) because
deque is implemented as a doubly linked list of blocks, designed for both-end operations.

Q6: When would you choose a generator expression over a list comprehension?

A: When you don't need all the results in memory at once — e.g., processing a huge file line by line, or when you'll only
consume part of the sequence (early exit).

# Industry-Level Q&A
Q7: You need to deduplicate 100 million records that don't fit in memory. A plain Python set won't work. What would
you reach for, conceptually?

A: Streaming/external approaches — sort-based dedup (external merge sort), a database with a unique index, or probabilistic
structures like a Bloom filter if approximate dedup is acceptable.

Q8: Why might dict.items() returning a "view" rather than a list matter in a multi-threaded context?

A: A view is live — if the underlying dict is mutated by another thread while you're iterating the view, you can get a
RuntimeError: dictionary changed size during iteration , or inconsistent reads. List snapshots avoid that at
the cost of memory and staleness.

Q9: How does Python's dict achieve O(1) average lookup internally?

A: It's a hash table — keys are hashed to bucket indices, so lookup, insert, and delete don't need to scan all entries. Collisions
are handled via open addressing/probing internally.

```python
class Cart:
    def __init__(self):
        self.items = {}  # product_id -> qty

    def add_item(self, product_id, qty):
        self.items[product_id] = self.items.get(product_id, 0) + qty

    def remove_item(self, product_id):
        self.items.pop(product_id, None)

    def get_total(self, price_lookup):
        total = 0

        for product_id, qty in self.items.items():
            price = price_lookup.get(product_id, 0)
            total += price * qty

        return total
```
