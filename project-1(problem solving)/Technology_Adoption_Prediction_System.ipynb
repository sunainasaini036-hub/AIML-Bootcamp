```python
import numpy as np
import pandas as pd
np.random.seed(45)
n_users=10000

data ={
      "User_ID": [f"USR{i:05d}" for i in range(1, n_users + 1)],

# Demographic Features
    "Gender": np.random.choice(["Male", "Female"], n_users),
    "Age": np.random.randint(18, 65, n_users),
    "Education_Level": np.random.choice(  ["High School", "Diploma", "Graduate", "Post Graduate"], n_users),
    "Occupation": np.random.choice(["Student", "Employee", "Business", "Freelancer"], n_users),
    "Annual_Income": np.random.randint(100000, 1500000, n_users),

# Technology Awareness Features
    "Internet_Usage_Hours": np.random.randint(1, 12, n_users),
    "Social_Media_Usage_Hours": np.random.randint(0, 8, n_users),
    "Previous_Tech_Adoption_Count": np.random.randint(0, 15, n_users),
    "Digital_Literacy_Score": np.random.randint(1, 101, n_users),

# Product Related Features
    "Product_Price": np.random.randint(500, 50000, n_users),
    "Ease_of_Use_Rating": np.random.randint(1, 11, n_users),
    "Product_Awareness_Score": np.random.randint(1, 101, n_users),
    "Feature_Satisfaction_Score": np.random.randint(1, 101, n_users),

# Marketing Features
    "Advertisement_Exposure": np.random.randint(0, 50, n_users),
    "Email_Campaign_Clicks": np.random.randint(0, 20, n_users),
    "Promotional_Offer_Used": np.random.choice([0, 1], n_users),

# Social Influence Features
    "Friend_Recommendations": np.random.randint(0, 10, n_users),
    "Online_Review_Score": np.random.randint(1, 6, n_users),
    "Social_Influence_Score": np.random.randint(1, 101, n_users),

# Behavioural Features
    "App_Login_Frequency": np.random.poisson(5, n_users),
    "Average_Session_Duration": np.random.randint(1, 120, n_users),
    "Feature_Usage_Count": np.random.randint(0, 50, n_users),

# Target Variable
    "Technology_Adoption_Status": np.random.choice(["Adopted", "Not Adopted"], n_users, p=[0.65, 0.35])
}
df = pd.DataFrame(data)
df
```
