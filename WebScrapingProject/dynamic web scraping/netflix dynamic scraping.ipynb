## Import Required Libraries


Import all required Python libraries.

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options

from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

import pandas as pd
import time
```

## Launch Chrome Browser


Initialize Chrome WebDriver.

```python
options = Options()

options.add_argument("--start-maximized")


# Uncomment for background execution
# options.add_argument("--headless=new")


driver = webdriver.Chrome(options=options)
```

## Open Netflix Top 10 Website


Navigate Selenium browser to Netflix Top 10 webpage.

```python
url = "https://www.netflix.com/tudum/top10"


driver.get(url)
```

## Wait for Page to Load


Wait until Netflix Top 10 table data appears.

```python
wait = WebDriverWait(driver,30)


wait.until(
    EC.presence_of_all_elements_located(
        (By.CSS_SELECTOR,"table tbody tr")
    )
)


time.sleep(5)
```

## Locate Netflix Movie Elements


Find all movie rows from Netflix Top 10 table.

```python
rows = driver.find_elements(By.CSS_SELECTOR, "table tbody tr")
print("Total Movies Found:",len(rows))
```

## Create Lists


Create lists to store extracted movie information.

```python
rank_list = []
title_list = []
poster_list = []
weeks_list = []
views_list = []
runtime_list = []
hours_list = []
```

## Extract Netflix Movie Information


Extract:

- Rank
- Title
- Poster
- Weeks in Top 10
- Views
- Runtime
- Hours Viewed

```python
for row in rows:


    # Rank

    try:
        rank = row.find_element(By.CSS_SELECTOR,'[data-uia="top10-table-row-title"] .rank').text.strip()
    except:
        rank="N/A"
    # Title

    try:
        title = row.find_element(
            By.CSS_SELECTOR,
            '[data-uia="top10-table-row-title"] button'
        ).text.strip()

    except:

        title="N/A"



    # Poster

    try:
        poster = row.find_element(
            By.CSS_SELECTOR,
            '[data-uia="top10-table-row-title"] img'
        ).get_attribute("src")

    except:

        poster=""



    # Weeks

    try:
        weeks = row.find_element(
            By.CSS_SELECTOR,
            '[data-uia="top10-table-row-weeks"]'
        ).text.strip()

    except:

        weeks="N/A"



    # Views

    try:
        views = row.find_element(
            By.CSS_SELECTOR,
            '[data-uia="top10-table-row-views"]'
        ).text.strip()

    except:

        views="N/A"



    # Runtime

    try:
        runtime = row.find_element(
            By.CSS_SELECTOR,
            '[data-uia="top10-table-row-runtime"]'
        ).text.strip()

    except:

        runtime=""



    # Hours

    try:
        hours = row.find_element(
            By.CSS_SELECTOR,
            '[data-uia="top10-table-row-hours"]'
        ).text.strip()

    except:

        hours=""



    rank_list.append(rank)
    title_list.append(title)
    poster_list.append(poster)
    weeks_list.append(weeks)
    views_list.append(views)
    runtime_list.append(runtime)
    hours_list.append(hours)
```

## Create DataFrame


Convert scraped Netflix data into a structured dataset.

```python
df = pd.DataFrame({

    "Rank": rank_list,
    "Title": title_list,
    "Poster": poster_list,
    "Weeks": weeks_list,
    "Views": views_list,
    "Runtime": runtime_list,
    "Hours Viewed": hours_list,
    "Link": url

})


df.head()
```
