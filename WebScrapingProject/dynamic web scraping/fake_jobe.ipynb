```python
jobsfake= soup.find_all( "div", class_='container')
len(jobsfake)
jobsfake
```

```python
print(soup.find("div",class_="container"))
```

```python
fake_data = []
```

```python
for job in jobsfake:
    title = job.find("h2", class_="title is-5")
    company = job.find("div", class_="content")
    location = job.find("p", class_="location")
    date_posted = job.find("p", class_="is-small has-text-grey")
    fake_data.append({  
    "Title": title.get_text(strip=True) if title else "N/A",
    "Company": company.get_text(strip=True) if company else "N/A",
    "Location": location.get_text(strip=True) if location else "N/A",
    "Date Posted": date_posted.get_text(strip=True) if date_posted else "N/A"
    })
```

```python
df = pd.DataFrame(fake_data)
df
```
