## Import Required Libraries



Import all required Python libraries.

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

import pandas as pd
import time
```

## Launch Chrome Browser



Initialize the Chrome WebDriver.

```python
url = "https://www.imdb.com/chart/top/"
driver = webdriver.Chrome()
driver.get(url)
```

## Wait for Page to Load



Pause execution for a few seconds to ensure the webpage loads completely.

```python
time.sleep(2)
```

## Verify Website Loading
Check that the correct webpage has loaded by displaying its title.

```python
print(driver.title)
```

# Locate Movie Elements


Locate all movie containers from the IMDb Top Movies page. Each movie container includes the title, rating, year,certificate, and duration, which will be extracted in the following steps.

```python
# Locate all movie elements
movies = driver.find_elements(By.CSS_SELECTOR,"li.ipc-metadata-list-summary-item")
print("Total Movies Found:", len(movies))
```

## Extract Movie data
Extract and display the data
 of each movie.

```python
titles=[]
ratings=[]
years=[]
durations=[]
certificates=[]

for movie in movies:

    # Title
    try:
        title = movie.find_element(
            By.CSS_SELECTOR,
            "h4.ipc-title__text"
        ).text
    except:
        title = "N/A"

    titles.append(title)

    # Rating
    try:
        rating = movie.find_element(
            By.CSS_SELECTOR,
            ".ipc-rating-star--rating"
        ).text
    except:
        rating = "N/A"

    ratings.append(rating)

    # Metadata inside THIS movie only
    meta = movie.find_elements(
        By.CSS_SELECTOR,
        ".ipc-inline-list__item"
    )

    try:
        years.append(meta[0].text)
    except:
        years.append("N/A")

    try:
        durations.append(meta[1].text)
    except:
        durations.append("N/A")

    try:
        certificates.append(meta[2].text)
    except:
        certificates.append("N/A")
```

## Create DataFrame


Combine the extracted lists into a structured Pandas DataFrame.

```python
df = pd.DataFrame({
    "Title": titles,
    "Rating": ratings,
    "Year": years,
    "Duration": durations,
    "Certificate": certificates
})
df
```

## Export Dataset

```python
df.to_csv("IMDb_Top_Movies.csv",index=False)
print("Dataset exported successfully!")
```
