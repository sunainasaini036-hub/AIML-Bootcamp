## Import Required Libraries



Import all required Python libraries.

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import time
```

## Launch Chrome Browser



Initialize the Chrome WebDriver.

```python
driver = webdriver.Chrome()
url= "https://books.toscrape.com/"
driver.get(url)
```

## Wait for Page to Load



Pause execution for a few seconds to ensure the webpage loads completely.

```python
time.sleep(2)
```

## Locate Book Elements


Find all book containers using their class name.

```python
books = driver.find_elements(By.CLASS_NAME, "product_pod")

print("Total Books Found:", len(books))
```

## Extract Book Information


Extract the title and price of every book.

```python
for book in books:
    title = book.find_element(By.TAG_NAME, "img").get_attribute("alt")
    price = book.find_element(By.CLASS_NAME, "price_color").text
    in_stock = book.find_element(By.CLASS_NAME, "instock").text.strip()

    print("Title:", title)
    print("Price:", price)
    print("In Stock:", in_stock)
    print("-" * 40)
```

## Create Lists

```python
titles = []
prices = []
stock_status = []
```

## Store Data

```python
for book in books:
    title = book.find_element(By.TAG_NAME, "img").get_attribute("alt")
    price = book.find_element(By.CLASS_NAME, "price_color").text
    in_stock = book.find_element(By.CLASS_NAME, "instock").text.strip()

    titles.append(title)
    prices.append(price)
    stock_status.append(in_stock)
```

## Create DataFrame

```python
import pandas as pd

df = pd.DataFrame({
    "Book Title": titles,
    "Price": prices
})

df.head()
```

## Export Dataset

```python
df.to_csv("books_dataset.csv", index=False)

print("Dataset exported successfully!")
```

## Close Browser

Close the browser after the scraping process is complete.

```python
driver.quit()
```
