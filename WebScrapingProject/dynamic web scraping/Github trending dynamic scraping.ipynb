## Import Required Libraries


Import all required Python libraries.

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

import pandas as pd
import time
```

## Launch Chrome Browser
Initialize Chrome WebDriver.

```python
options = Options()

options.add_argument("--start-maximized")

# Uncomment for background execution
# options.add_argument("--headless=new")


driver = webdriver.Chrome(options=options)
```

## Open GitHub Trending Website
Open GitHub Trending page using Selenium

```python
url = "https://github.com/trending"

driver.get(url)
```

## Wait for Page to Load
Wait until repository elements are visible on the webpage.

```python
wait = WebDriverWait(driver,20)

wait.until(
    EC.presence_of_all_elements_located(
        (By.CSS_SELECTOR,"article.Box-row")
    )
)
```

## Locate Repository Elements
Find all trending repository containers.

```python
repositories = driver.find_elements(
    By.CSS_SELECTOR,
    "article.Box-row"
)


print("Total Repositories Found:", len(repositories))
```

## Extract Repository Information
Extract repository name, description, language, stars, forks and URL.

```python
for repo in repositories:

    title = repo.find_element(
        By.CSS_SELECTOR,
        "h2 a"
    ).text

    link = repo.find_element(
        By.CSS_SELECTOR,
        "h2 a"
    ).get_attribute("href")

    print("Repository:", title)
    print("Link:", link)
    print("-"*40)
```

## Create Lists

```python
titles = []
descriptions = []
languages = []
stars = []
forks = []
links = []
```

## Store Data
Store extracted repository information into lists.

```python
for repo in repositories:

    try:
        title = repo.find_element(
            By.CSS_SELECTOR,
            "h2 a"
        ).text.replace("\n"," ")

    except:
        title="N/A"



    try:
        link = repo.find_element(
            By.CSS_SELECTOR,
            "h2 a"
        ).get_attribute("href")

    except:
        link="N/A"



    try:
        description = repo.find_element(
            By.CSS_SELECTOR,
            "p"
        ).text

    except:
        description="No Description"



    try:
        language = repo.find_element(
            By.CSS_SELECTOR,
            "[itemprop='programmingLanguage']"
        ).text

    except:
        language="N/A"



    try:
        stats = repo.find_elements(
            By.CSS_SELECTOR,
            "a.Link--muted"
        )

        stars_value = stats[0].text
        forks_value = stats[1].text

    except:

        stars_value="N/A"
        forks_value="N/A"



    titles.append(title)
    descriptions.append(description)
    languages.append(language)
    stars.append(stars_value)
    forks.append(forks_value)
    links.append(link)
```

## Create DataFrame
Convert scraped data into a structured dataset.

```python
df = pd.DataFrame({

    "Repository Name": titles,
    "Description": descriptions,
    "Language": languages,
    "Stars": stars,
    "Forks": forks,
    "Repository Link": links

})
df
```

Save scraped repository data into CSV format.

```python
df.to_csv("github_trending_dataset.csv",index=False)
print("Dataset exported successfully!")
```
