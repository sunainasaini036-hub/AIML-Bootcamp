## Import Required Libraries


Import all required Python libraries.

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options

from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

import pandas as pd
import time
```

## Launch Chrome Browser


Initialize Chrome WebDriver with browser settings.

```python
options = Options()
options.add_argument("--start-maximized")
options.add_argument("user-agent=Mozilla/5.0")
driver = webdriver.Chrome(options=options)
```

## Open YouTube Trending Page


Navigate Selenium browser to YouTube search results.

```python
url = "https://www.youtube.com/results?search_query=trending+videos"


driver.get(url)
```

## Wait for Videos to Load


Wait until YouTube video elements appear.

```python
wait = WebDriverWait(driver,20)


wait.until(
    EC.presence_of_element_located(
        (By.TAG_NAME,"ytd-video-renderer")
    )
)
```

## Scroll Page


Scroll multiple times to load additional videos dynamically.

```python
for i in range(8):

    driver.execute_script(
        "window.scrollTo(0, document.documentElement.scrollHeight);"
    )

    time.sleep(2)
```

## Locate Video Containers


Find all available YouTube video containers.

```python
videos = driver.find_elements(
    By.TAG_NAME,
    "ytd-video-renderer"
)


print("Videos Found:",len(videos))
```

## Create Lists


Create lists to store extracted video information.

```python
titles = []
images = []
channels = []
views = []
durations = []
links = []
```

## Extract YouTube Video Information


Extract:

- Video Title
- Thumbnail
- Channel Name
- Views
- Duration
- Video Link

```python
seen = set()


for video in videos:


    try:

        title_element = video.find_element(
            By.CSS_SELECTOR,
            "a#video-title"
        )


        title = title_element.text.strip()

        link = title_element.get_attribute("href")


        if not title or title in seen:
            continue


        seen.add(title)


    except:

        continue



    # Thumbnail

    try:

        img = video.find_element(
            By.CSS_SELECTOR,
            "img"
        )


        image = (
            img.get_attribute("src")
            or img.get_attribute("data-thumb")
            or ""
        )


    except:

        image=""



    # Channel

    try:

        channel = video.find_element(
            By.CSS_SELECTOR,
            "#channel-name a"
        ).text.strip()


    except:

        channel="N/A"



    # Views

    try:

        metadata = video.find_elements(
            By.CSS_SELECTOR,
            "#metadata-line span"
        )


        views_value = (
            metadata[0].text
            if len(metadata)>0
            else "N/A"
        )


    except:

        views_value="N/A"



    # Duration

    try:

        duration = video.find_element(
            By.CSS_SELECTOR,
            "span.ytd-thumbnail-overlay-time-status-renderer"
        ).text.strip()


    except:

        duration="N/A"



    titles.append(title)
    images.append(image)
    channels.append(channel)
    views.append(views_value)
    durations.append(duration)
    links.append(link)



    if len(titles)==20:
        break
```

## Create DataFrame


Convert scraped YouTube data into structured format.

```python
df = pd.DataFrame({

    "Video Title": titles,
    "Thumbnail": images,
    "Channel": channels,
    "Views": views,
    "Duration": durations,
    "Video Link": links

})


df.head()
```

Save YouTube video data into CSV format.
