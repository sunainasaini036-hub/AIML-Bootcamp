## Web Scraping

### How Web Scraping is done?
  - Sending an HTTP request : sent to the target web server to retrive the desired web page.
  - Retrieving the Web Page Content: Server responds with the web page content, usually HTML format.
  - Parsing the HTML: HTML content is parsed to locate the specific data elements using libraries or tools designed for this purpose.
[Beautifulsoup, Scrapy]
  - Extracting Data: Relevant data is extracted from the parsed HTML structure based on the predefined rules or criteria.
  - Storing Data: The extracted data can be stored in various format such as csv, json or database, for further analysis.
  - Handling Challenges: Web Scrapers may need to address challenges like
    - Anti-scraping measures
    - Dynamic content loading (JavasScript)
    - Legal considerations

```python
products = soup.find_all("div", {"data-component-type": "s-search-result"})

links_list = []

for product in products:
    a_tag = product.find("a", class_="a-link-normal")
    
    if a_tag:
        href = a_tag.get("href")

        if href and href.startswith("/"):
            full_link = "https://www.amazon.in" + href
            links_list.append(full_link)
```
