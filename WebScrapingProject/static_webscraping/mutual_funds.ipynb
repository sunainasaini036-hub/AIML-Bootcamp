# Import Required Libraries

Import all the libraries needed for web scraping, data handling, and exporting the scraped data.

```python
from bs4 import BeautifulSoup
import requests
import pandas as pd
import numpy as np
```

# Define the Target Website

Specify the URL of the webpage that will be scraped.

```python
url = "https://finance.yahoo.com/markets/mutualfunds/top/"
response = requests.get(url)
```

# Add Browser Headers


Provide request headers so the website recognizes the request as coming from a real web browser.

```python
HEADERS = ({'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/126.0.0.0 Safari/537.36'

})
```

# Send HTTP Request

Send a GET request to the website and retrieve its HTML content.

```python
response = requests.get(url,headers = HEADERS)
```

# Check the Response


Verify whether the request was successful by checking the HTTP status code.

```python
print(response.status_code)
```

# Inspect Response Content


Check the type of the downloaded content.

```python
type(response.content)
```

# Parse the HTML

Convert the HTML into a BeautifulSoup object for easier navigation.

```python
soup = BeautifulSoup(response.content,"html.parser")
```

# Preview the HTML Structure


Display the parsed HTML structure for understanding the webpage layout.

```python
soup
```

```python
print(soup.prettify())
```

# Locate Quote Containers


Find all quote cards using their HTML class.

```python
mutual_funds= soup.find_all( "div", class_='container yf-1nql2a7')
len(mutual_funds)
mutual_funds
```

# Verify Scraped Elements


Display the first quote container to understand its HTML structure.

```python
print(soup.find("div",class_="container yf-1nql2a7"))
```

# Display All Quote Containers


Print all the quote elements extracted from the webpage.

```python
print(mutual_funds)
```

# Count Total Quotes


Determine the total number of quotes available on the page.

```python
print(len(mutual_funds))
```

##   Extract Mutual Fund Data


Extract the required mutual fund information from the HTML elements and store it in Python lists for further processing.

```python
data = []
```

```python
rows= soup.find_all("tr")
for row in rows:
    cols= row.find_all("td")
    data.append([col.get_text(strip=True) for col in cols])
```

## Create DataFrame


Combine the extracted lists into a structured Pandas DataFrame.

```python
df = pd.DataFrame(data)
df
```

## Inspect Dataset


Verify that the extracted data is correctly stored.

```python
df.head()
```

## Export Dataset


Save the cleaned dataset for future use.

```python
df.to_csv("scraped_mutual_data.csv", index=False)
print("CSV file created successfully!")
```
