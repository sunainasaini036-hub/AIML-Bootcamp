## Import Required Libraries
Import all required Python libraries.

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import numpy as np
import re
```

## Define Wikipedia URL

Provide the Wikipedia page URL from which data will be extracted

```python
url = "https://en.wikipedia.org/wiki/Albert_Einstein"
```

# Add Browser Headers


Provide request headers so the website recognizes the request as coming from a real web browser.

```python
HEADERS = ({'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/126.0.0.0 Safari/537.36'

})
```

## Send HTTP Request


Connect with Wikipedia server and fetch HTML content.

```python
response = requests.get(url,headers = HEADERS)
```

## Check the Response


Verify whether the request was successful by checking the HTTP status code.

```python
print(response.status_code)
if response.status_code == 200:
    print("Website connected successfully")
else:
    print("Failed:", response.status_code)
```

## Check Response Content Type

Inspect the returned webpage data.

```python
type(response.content)
```

# Parse the HTML

Convert the HTML into a BeautifulSoup object for easier navigation.

```python
soup = BeautifulSoup(response.content,"html.parser")
```

## Preview HTML Structure

Display webpage HTML structure.

```python
soup
```

## Extract Wikipedia Title

Find the main heading of the Wikipedia page.

```python
title = soup.find("h1")

print(title.text)
```

## Extract Article Paragraphs

Extract the main text content from Wikipedia.

```python
paragraphs = soup.find_all("p")

len(paragraphs)
```

## Display Paragraph Data

View extracted article content.

```python
for para in paragraphs[:5]:
    print(para.text)
```

## Store Article Data

Create a list containing all paragraphs.

```python
article_data = []


for para in paragraphs:

    text = para.text.strip()

    if text:
        article_data.append(text)


print(len(article_data))
```

## Create Article DataFrame

Convert article text into structured format.

```python
article_df = pd.DataFrame({

"Article_Content":article_data

})


article_df.head()
```

## Extract Wikipedia Headings

Find all section headings.

```python
headings = []


for heading in soup.find_all(["h2","h3"]):

    headings.append(
        heading.text.strip()
    )


headings[:10]
```

## Extract Infobox Data

Wikipedia pages contain important information inside infobox tables.

```python
infobox = soup.find("table",class_="infobox")
print(infobox)
```

## Extract Infobox Information

```python
infobox_data = {}


if infobox:

    rows = infobox.find_all("tr")

    for row in rows:

        cells = row.find_all(["th","td"])

        if len(cells)==2:

            key = cells[0].text.strip()
            value = cells[1].text.strip()

            infobox_data[key]=value

else:
    print("No Infobox Found")


infobox_data
```

## Extract Wikipedia Tables

Find all tables available on the webpage.

```python
tables = soup.find_all("table")

print(
"Total Tables:",
len(tables)
)
```

## Extract Wikipedia Table Data

```python
table_data = []


for row in tables[0].find_all("tr"):

    cells = row.find_all(["th","td"])

    row_data = []

    for cell in cells:
        row_data.append(cell.text.strip())


    if row_data:
        table_data.append(row_data)


table_df = pd.DataFrame(table_data)


table_df.head()
```

# Export to CSV


Save the scraped data as a CSV file for future use.

```python
table_df.to_csv(
    "wikipedia_table_data.csv",
    index=False
)

print("CSV created successfully!")
```
