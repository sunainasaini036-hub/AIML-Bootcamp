## Import Required Libraries
Import essential Python libraries for API requests and data processing.

```python
import requests
import pandas as pd
```

## Define API Endpoint
Specify the Universities API URL to retrieve Indian university data.

```python
url ="http://universities.hipolabs.com/search?country=India"

response=requests.get(url)
```

## Send API Request
Establish a connection with the REST API using an HTTP GET request.

```python
response = requests.get(url)
```

## Validate API Response
Verify the request was successful by checking the HTTP status code (200 OK).

```python
print(response.status_code)
```

## Retrieve JSON Data
Fetch and display the JSON response received from the API.

```python
print(response.json())
```

## Store API Response
Save the JSON data into a Python variable for further processing.

```python
data = response.json()
```

## Iterate Through Records
Loop through each university record returned by the API.

```python
for item in data:
    print(item)
```

## Extract Required Information
Access fields such as university name, website, and domain.

```python
for item in data:
    print("University Name :", item["name"])
    print("Website         :", item["web_pages"][0])
    print("Domain          :", item["domains"][0])
    print("-" * 50)
```

## Create Structured Dataset
Organize the extracted data into a structured format.

```python
universities = []

for item in data:
    universities.append({
        "University Name": item["name"],
        "Website": item["web_pages"][0],
        "Domain": item["domains"][0],
        "Country": item["country"],
        "State": item["state-province"]
    })

print(universities)
```

## Convert to Pandas DataFrame
Transform the data into a DataFrame for analysis.

```python
df = pd.DataFrame(universities)
df
```

## Preview Dataset
Display the first few rows of the dataset.

```python
df.head()
```

## Save DataFrame as a CSV file

```python
df.to_csv("Indian_Universities.csv", index=False)

print("CSV file saved successfully!")
```
