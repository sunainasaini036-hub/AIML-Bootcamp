## Import Required Libraries
Import essential Python libraries for API requests and data processing.

```python
import requests
import pandas as pd
```

## Define Currency Exchange API Endpoint
Specify the Exchange Rate API URL to retrieve real-time currency exchange data.

```python
url="https://api.frankfurter.app/latest?from=USD&to=INR,EUR,GBP,JPY"
```

## Send API Request
Establish a connection with the REST API using an HTTP GET request.

```python
response = requests.get(url)
```

## Validate API Response
Verify the request was successful by checking the HTTP status code (200 OK).

```python
print(response.status_code)
```

## Retrieve JSON Exchange Rate Data
Fetch and display the JSON response returned by the Currency Exchange API.

```python
currency_data = response.json()

currency_data
```

## Store API Response Data
Store the API response in a Python variable for further processing and analysis.

```python
exchange_data = currency_data

exchange_data
```

## Explore API Response Structure
Examine the JSON response to understand its keys, base currency, date, and exchange rates.

```python
print("Available Keys:")
print(exchange_data.keys())

print("\nBase Currency:", exchange_data["base"])

print("Date:", exchange_data["date"])
```

## Extract Currency Rate Information
Loop through each university record returned by the API.

```python
exchange_rates = exchange_data["rates"]

exchange_rates
```

## Create Structured Currency Dataset
Organize the extracted currency information into a structured list of dictionaries.

```python
currency_list = []

for currency, rate in exchange_rates.items():
    currency_list.append({
        "Currency": currency,
        "Exchange Rate": rate
    })

currency_list
```

## Convert to Pandas DataFrame
Transform the structured data into a DataFrame for analysis.

```python
currency_df = pd.DataFrame(currency_list)

currency_df
```

## Preview Currency Dataset
Display the first few records of the processed currency exchange dataset.

```python
currency_df.head()
```

## Filter Required Currency Rates
Filter and display selected currencies for further analysis or conversion.

```python
selected_currency = currency_df[
    currency_df["Currency"].isin(["INR", "EUR", "GBP", "JPY"])
]

selected_currency
```

## Create Currency Conversion Function
Develop a reusable function to perform currency conversion using real-time exchange rates.

```python
def currency_converter(amount, target_currency):
    exchange_rate = exchange_rates[target_currency]
    converted_amount = amount * exchange_rate
    return converted_amount
```

## Perform Currency Conversion
Convert a user-specified amount between different currencies using the latest exchange rates.

```python
amount = 100

print(f"100 USD = {currency_converter(amount,'INR'):.2f} INR")
print(f"100 USD = {currency_converter(amount,'EUR'):.2f} EUR")
print(f"100 USD = {currency_converter(amount,'GBP'):.2f} GBP")
print(f"100 USD = {currency_converter(amount,'JPY'):.2f} JPY")
```

## Export Currency Dataset
Save the processed currency exchange dataset as a CSV file for future use.

```python
currency_df.to_csv("currency_exchange_rates.csv", index=False)

print("Dataset exported successfully.")
```
