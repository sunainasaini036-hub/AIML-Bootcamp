**1. PROBLEM OBJECTIVE AND DATASET DESCRIPTION**

Activation functions are one of the most important components of an artificial neural network. They determine how the output of a neuron is transformed before being passed to the next layer.

Without an appropriate activation function, multiple layers of a neural network would behave almost like a single linear transformation. Non-linear activation functions allow the network to learn complicated patterns and decision boundaries from data.

The objective of this experiment is to compare four commonly used activation functions:


1.Identity / Linear

2.Sigmoid / Logistic

3.Hyperbolic Tangent / Tanh

4.ReLU

The comparison is performed using a neural-network classifier trained on the Breast Cancer Wisconsin (Diagnostic) dataset.


Dataset Overview

**2. DATA PREPROCESSING**

Before training the neural networks, the input features were standardized.

Standardization transforms every feature according to:

$$ z=\frac{x-\mu}{\sigma} $$

where:

\(x\) = original feature value

\(\mu\) = mean of the feature

\(\sigma\) = standard deviation


Feature scaling is particularly useful for neural networks because the input variables have different numerical ranges.


```python
import numpy as np
import pandas as pd

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Load dataset
data = load_breast_cancer()

X = data.data
y = data.target

# Split data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=42
)

# Standardization
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

print("Training Samples:", X_train.shape[0])
print("Testing Samples:", X_test.shape[0])
print("Number of Features:", X_train.shape[1])
```

**4. IMPLEMENTATION OF DIFFERENT ACTIVATION FUNCTIONS**

```python
from sklearn.neural_network import MLPClassifier

activations = [
    "identity",
    "logistic",
    "tanh",
    "relu"
]

models = {}

for activation in activations:

    model = MLPClassifier(
        hidden_layer_sizes=(64, 32),
        activation=activation,
        solver="adam",
        max_iter=500,
        random_state=42
    )

    model.fit(X_train, y_train)

    models[activation] = model

print("Training completed for all activation functions.")
```

**5. ACTIVATION FUNCTION 1 — IDENTITY**

The Identity activation function is defined as:

$$ f(x)=x $$

Its output is exactly equal to its input.

Characteristics

Very simple mathematical operation.

Does not introduce non-linearity.

Computationally inexpensive.

Useful when a linear relationship is appropriate.

Not generally suitable for hidden layers when complex patterns need to be learned.

Important Point

If all hidden layers use an identity activation, the combination of several layers can still be represented as one overall linear transformation.

Therefore, increasing the number of hidden layers does not provide the same expressive power as a network containing non-linear activations.

**6. ACTIVATION FUNCTION 2 — SIGMOID / LOGISTIC**

The Sigmoid activation function is:

$$ f(x)=\frac{1}{1+e^{-x}} $$

Its output lies between 0 and 1.

Characteristics

The sigmoid curve is smooth and gradually changes from values close to 0 to values close to 1.

It can be useful when an output needs to represent a probability-like value.

Advantages

Smooth and differentiable.

Output is bounded between 0 and 1.

Easy to interpret as a probability in suitable output-layer applications.

Limitations

For very large positive or negative input values, the function becomes almost flat.

Consequently, its gradient becomes very small. This can slow down learning in deeper neural networks.

Another limitation is that sigmoid outputs are not centered around zero.

**7. ACTIVATION FUNCTION 3 — TANH**

The Hyperbolic Tangent activation function is:

$$ f(x)=\tanh(x) $$

It produces values approximately between:

$$ -1 \text{ and } +1 $$
Characteristics

Unlike sigmoid, Tanh is zero-centered.

This can make optimization more balanced because positive and negative activations are both possible.

Advantages

Zero-centered output.

Smooth and differentiable.

Can model non-linear relationships.

Often works well for smaller neural networks.

Limitations

Tanh can also suffer from the vanishing-gradient problem when its input becomes very large or very small.

Therefore, although it introduces useful non-linearity, it may not always train as efficiently as ReLU in larger deep networks.

8. ACTIVATION FUNCTION 4 — RELU

The Rectified Linear Unit is defined as:

$$ f(x)=\max(0,x) $$

This means:

If \(x<0\), output = 0

If \(x>0\), output = \(x\)

Characteristics

ReLU is computationally simple and introduces non-linearity without requiring an expensive mathematical operation such as an exponential function.

Advantages
Simple and fast.
Usually provides efficient gradient propagation for positive inputs.
Works particularly well in many modern neural-network architectures.
Produces sparse activations because negative values become zero.
Limitation

ReLU can produce a dying ReLU problem.

If a neuron repeatedly receives negative inputs, its output can remain zero and the neuron may stop contributing effectively to learning.

**9. COMPARISON OF ACTIVATION FUNCTIONS**

Tanh	\(\tanh(x)\)	(-1,1)	Zero-centered non-linearity

ReLU	\(\max(0,x)\)	[0,+∞)	Simple and efficient non-linearity

**10. MODEL EVALUATION**

The trained models were evaluated using:

Accuracy

$$ Accuracy=\frac{Correct\ Predictions}{Total\ Predictions} $$

Accuracy represents the percentage of test samples classified correctly.

Weighted F1 Score

F1 score combines precision and recall:

$$ F1=2\frac{Precision\times Recall}{Precision+Recall} $$

Weighted F1 accounts for the number of samples belonging to each class.

Training Iterations

The number of iterations required by the model before training stopped was also recorded.

Final Training Loss

The final loss value gives an indication of the optimization state reached by the model on the training data.

**12. PYTHON CODE FOR PERFORMANCE COMPARISON**

```python

from sklearn.metrics import accuracy_score, f1_score

results = []

for activation, model in models.items():

    predictions = model.predict(X_test)

    accuracy = accuracy_score(y_test, predictions)

    f1 = f1_score(
        y_test,
        predictions,
        average="weighted"
    )

    results.append({
        "Activation": activation,
        "Accuracy": accuracy,
        "Weighted F1": f1,
        "Iterations": model.n_iter_,
        "Training Loss": model.loss_
    })

results_df = pd.DataFrame(results)

print(results_df)
```

**13. CONFUSION MATRIX**

A confusion matrix can be used to examine the classification behavior of each model.



```python

import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

for activation, model in models.items():

    predictions = model.predict(X_test)

    ConfusionMatrixDisplay.from_predictions(
        y_test,
        predictions
    )

    plt.title(
        f"Confusion Matrix - {activation.title()}"
    )

    plt.show()
```

**Purpose of the Graph**

The confusion matrix shows:

•Correctly classified malignant samples

•Correctly classified benign samples

•Malignant samples incorrectly classified as

•Benign samples incorrectly classified as malignant

This provides more information than accuracy alone.



**14. LOSS CONVERGENCE ANALYSIS**

The training loss curve can be plotted for every activation function.


```python

plt.figure(figsize=(10,6))

for activation, model in models.items():

    plt.plot(
        model.loss_curve_,
        label=activation.title()
    )

plt.xlabel("Training Iterations")
plt.ylabel("Loss")
plt.title("Loss Convergence for Different Activation Functions")
plt.legend()
plt.grid(True)

plt.show()
```


**Interpretation**

The loss graph allows us to observe how quickly each model learns.

**In this experiment:**

•ReLU reached the stopping condition in 164 iterations.

•Identity required 211 iterations.

•Sigmoid required 278 iterations.

•Tanh required 306 iterations.

Thus, under this particular configuration, ReLU showed the fastest convergence.
