### 4. What is a mutable default argument bug? How do you fix it?

Anwer: In Python, default argument values are created only once, when the function is defined, not each time it is called.


If the default value is mutable (like a list, dict, or set), changes made in one call will be remembered in future calls.

**Bug Example**

```python
def add_item(item, my_list=[]):
    my_list.append(item)
    return my_list

print(add_item(1))
print(add_item(2))
print(add_item(3))
```

# How to Fix It
Use None as the default value and create a new list inside the function.

```python
def add_item(item, my_list=None):
    if my_list is None:
        my_list = []

    my_list.append(item)
    return my_list

print(add_item(1))
print(add_item(2))
print(add_item(3))
```

### 10. What does @functools.wraps(func) do, and why is it mandatory in decorators?

    @functools.wraps(func) is used inside a decorator to preserve the original function's information (metadata) when it is wrapped by another function.

    Without @wraps, the decorated function looks like the wrapper function instead of the original one.

    Example Without @wraps

```python
def decorator(func):
    def wrapper():
        return func()
    return wrapper

@decorator
def greet():
    """Say Hello"""
    print("Hello")

print(greet.__name__)
```

The original function name greet is lost.

Example With @wraps

```python
from functools import wraps

def decorator(func):
    @wraps(func)
    def wrapper():
        return func()
    return wrapper

@decorator
def greet():
    """Say Hello"""
    print("Hello")

print(greet.__name__)
```

Now the original function information is preserved.

# EC-1: EMPTY INPUT CASES
# The function below works for normal input — make it handle all edge cases


```python
def average(numbers):
 """Return the average of a list of numbers."""
 return sum(numbers) / len(numbers)
```

# Test cases to handle:
# average([]) → should return None or raise a clear ValueError
# average([5]) → should return 5.0
# average(None) → should raise TypeError with clear message
# average(["a", "b"]) → should raise TypeError
# average([1, 2, None, 4]) → handle None in list
# Your improved version:


```python
def average(numbers):
    """
    Return the average of a list of numbers.
    """

    # Check for None input
    if numbers is None:
        raise TypeError("Input cannot be None.")

    # Check if input is a list
    if not isinstance(numbers, list):
        raise TypeError("Input must be a list of numbers.")

    # Check for empty list
    if len(numbers) == 0:
        raise ValueError("Cannot calculate average of an empty list.")

    # Validate all elements
    for num in numbers:
        if num is None:
            raise TypeError("List contains None value.")
        if not isinstance(num, (int, float)):
            raise TypeError("All elements must be numbers.")

    return sum(numbers) / len(numbers)
```

def find_duplicates(items):
    """Find items that appear more than once."""

    seen = set()
    duplicates = set()

    for item in items:
        if item in seen:
            duplicates.add(item)
        else:
            seen.add(item)

    return list(duplicates)
reason:Set lookups (item in seen) are O(1) on average.
Each element is processed only once.Total time complexity becomes O(n).
Space complexity remains O(n) because of the sets.

```python
# Test your implementation:
test_cases = [
 (50, "free"), # Below ₹100
 (100, 2.0), # Minimum fee
 (500, 9.0), # Normal: 1.8% of 500
 (70000, 1000.0), # Maximum fee cap
 (499.99, None), # Float input
 ("500", None), # String input
 (-100, None), # Negative — invalid
 (0, None), # Zero — invalid
]
for amount, expected in test_cases:
 result = calculate_upi_fee(amount)
 status = " " if result == expected or (expected is None and result is None) else " "
 print(f"{status} calculate_upi_fee({amount}) = {result} | expected: {expected}")
```

```python
def calculate_upi_fee(amount):
    """
     Calculate UPI transaction fee.

     Returns:
float: fee amount
None: if input is invalid
    """

    try:
        # Convert string inputs like "500" to float
        amount = float(amount)

        # Reject zero or negative amounts
        if amount <= 0:
            print("Error: Transaction amount must be greater than 0.")
            return None

        # Transactions below ₹100 are free
        if amount < 100:
            return 0.0

        # Calculate 1.8% fee
        fee = amount * 0.018

        # Apply min/max limits
        fee = max(2, fee)
        fee = min(1000, fee)

        # Round to 2 decimal places
        return round(fee, 2)

    except (TypeError, ValueError):
        print(f"Error: Invalid transaction amount: {amount}")
        return None
```

```python
def calculate_bmi(weight_kg, height_m):
    """
    Calculate BMI and classify health risk.

    Returns:
        dict with bmi, category, risk_level
        None if input is invalid
    """

    try:
        # Convert inputs to float (handles numeric strings like "70")
        weight_kg = float(weight_kg)
        height_m = float(height_m)

        # Validate ranges
        if not (1 <= weight_kg <= 500):
            print("Error: Weight must be between 1 and 500 kg.")
            return None

        if not (0.5 <= height_m <= 2.5):
            print("Error: Height must be between 0.5 and 2.5 meters.")
            return None

        # Calculate BMI
        bmi = round(weight_kg / (height_m ** 2), 1)

        # Classify BMI and risk level
        if bmi < 18.5:
            category = "Underweight"
            risk_level = "Moderate"
        elif bmi < 25:
            category = "Normal"
            risk_level = "Low"
        elif bmi < 30:
            category = "Overweight"
            risk_level = "Elevated"
        else:
            category = "Obese"
            risk_level = "High"

        return {
            "bmi": bmi,
            "category": category,
            "risk_level": risk_level
        }

    except (TypeError, ValueError):
        print("Error: Weight and height must be numeric values.")
        return None


# Test cases
print(calculate_bmi(70, 1.75))
print(calculate_bmi(50, 1.75))
print(calculate_bmi(100, 1.75))
print(calculate_bmi(-5, 1.75))
print(calculate_bmi(70, 0))
print(calculate_bmi("seventy", 1.75))
```

```python
def apply_discount(price, coupon_code):
    """
    Apply discount coupon to a cart price.

    Returns dict with original, discount, final prices, and a message.
    """

    try:
        price = float(price)

        if price <= 0:
            return {
                "original": price,
                "discount": 0.0,
                "final": price,
                "message": "Invalid price."
            }

        coupon_code = str(coupon_code).upper()

        # Minimum order amount check
        if price < 200:
            return {
                "original": price,
                "discount": 0.0,
                "final": price,
                "message": "Order must be at least ₹200 for discounts."
            }

        discount = 0.0

        if coupon_code == "SAVE10":
            discount = price * 0.10

        elif coupon_code == "SAVE20":
            discount = price * 0.20

        elif coupon_code == "FLAT50":
            if price > 500:
                discount = 50.0
            else:
                return {
                    "original": price,
                    "discount": 0.0,
                    "final": price,
                    "message": "FLAT50 requires order value above ₹500."
                }

        else:
            return {
                "original": price,
                "discount": 0.0,
                "final": price,
                "message": "Invalid coupon code."
            }

        # Maximum discount = 30% of original price
        max_discount = price * 0.30
        discount = min(discount, max_discount)

        # Final price cannot be negative
        final_price = max(0, price - discount)

        return {
            "original": round(price, 2),
            "discount": round(discount, 2),
            "final": round(final_price, 2),
            "message": "Discount applied successfully."
        }

    except (TypeError, ValueError):
        return {
            "original": 0.0,
            "discount": 0.0,
            "final": 0.0,
            "message": "Invalid input."
        }
```

```python
def get_hra_rate(city):
    metro_cities = {
        "mumbai", "delhi", "bangalore",
        "chennai", "kolkata", "hyderabad"
    }

    if str(city).lower() in metro_cities:
        return 0.50
    return 0.40


def calculate_tax(annual_income):
    if annual_income <= 250000:
        return 0
    elif annual_income <= 500000:
        return annual_income * 0.05
    elif annual_income <= 1000000:
        return annual_income * 0.20
    else:
        return annual_income * 0.30


def calculate_net_salary(basic_monthly, city="non-metro"):
    if basic_monthly <= 0:
        return None

    hra = basic_monthly * get_hra_rate(city)
    gross = basic_monthly + hra

    pf = basic_monthly * 0.12 if basic_monthly >= 15000 else 0
    pt = 200 if basic_monthly > 15000 else 0

    annual_tax = calculate_tax(gross * 12)

    net = gross - pf - pt - (annual_tax / 12)

    return {
        "gross_salary": round(gross, 2),
        "net_salary": round(net, 2)
    }
```

```python
import math

VALID_ZONES = {"local", "regional", "national", "international"}

ZONE_SURCHARGE = {
    "local": 0,
    "regional": 20,
    "national": 50,
    "international": 200
}


def calculate_shipping(
    weight_g,
    zone,
    express=False,
    fragile=False,
    insurance_value=0
):
    """
    Calculate total shipping cost.

    Returns dict with itemized cost breakdown.
    """

    try:
        weight_g = float(weight_g)

        if weight_g <= 0:
            raise ValueError("Weight must be positive.")

        zone = zone.lower()

        if zone not in VALID_ZONES:
            raise ValueError(
                f"Invalid zone. Must be one of {VALID_ZONES}"
            )

        # Base cost calculation
        if weight_g <= 500:
            base_cost = 50
        else:
            extra_blocks = math.ceil((weight_g - 500) / 500)
            base_cost = 50 + (extra_blocks * 30)

        zone_charge = ZONE_SURCHARGE[zone]

        express_charge = 0
        if express:
            express_charge = 0.5 * base_cost

        fragile_charge = 25 if fragile else 0

        insurance_charge = 0.02 * insurance_value

        total = (
            base_cost
            + zone_charge
            + express_charge
            + fragile_charge
            + insurance_charge
        )

        return {
            "base_cost": round(base_cost, 2),
            "zone_surcharge": round(zone_charge, 2),
            "express_charge": round(express_charge, 2),
            "fragile_charge": round(fragile_charge, 2),
            "insurance_charge": round(insurance_charge, 2),
            "total_cost": round(total, 2)
        }

    except (ValueError, TypeError) as e:
        print(f"Error: {e}")
        return None
```

```python
import functools
import time

def retry(max_attempts=3, delay=0):
    def decorator(func):
        @functools.wraps(func)
        def wrapper(*args, **kwargs):
            last_exception = None

            for attempt in range(max_attempts):
                try:
                    return func(*args, **kwargs)
                except Exception as e:
                    last_exception = e

                    # Wait before retrying (except after last attempt)
                    if attempt < max_attempts - 1:
                        time.sleep(delay)

            # All attempts failed
            raise last_exception

        return wrapper
    return decorator

attempt_count = 0

@retry(max_attempts=3)
def flaky_api_call():
    global attempt_count
    attempt_count += 1

    if attempt_count < 3:
        raise ConnectionError("API timeout")

    return "success"

print(flaky_api_call())  # Should succeed on 3rd attempt
print(f"Took {attempt_count} attempts")
```

```python
import functools

def memoize(func):
    cache = {}
    hits = 0
    misses = 0

    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        nonlocal hits, misses

        # Create a hashable key from arguments
        key = (args, tuple(sorted(kwargs.items())))

        if key in cache:
            hits += 1
            return cache[key]

        misses += 1
        result = func(*args, **kwargs)
        cache[key] = result
        return result

    def cache_info():
        return {
            "hits": hits,
            "misses": misses,
            "size": len(cache)
        }

    wrapper.cache_info = cache_info
    return wrapper


@memoize
def fib(n):
    if n < 2:
        return n
    return fib(n - 1) + fib(n - 2)


print(fib(30))          # Should be fast
print(fib.cache_info()) # Should show hits, misses, size
```

```python
def has_duplicate_slow(items):
    for i in range(len(items)):
        for j in range(i + 1, len(items)):
            if items[i] == items[j]:
                return True
    return False


# O(n) solution using a set
def has_duplicate_fast(items):
    seen = set()

    for item in items:
        if item in seen:
            return True
        seen.add(item)

    return False


# Verify same results
assert has_duplicate_fast([1, 2, 3]) == has_duplicate_slow([1, 2, 3])
assert has_duplicate_fast([1, 2, 2, 3]) == has_duplicate_slow([1, 2, 2, 3])

# Time both
import time

test_data = list(range(10000)) + [9999]  # duplicate at end

start = time.time()
has_duplicate_slow(test_data)
print(f"Slow: {time.time() - start:.4f}s")

start = time.time()
has_duplicate_fast(test_data)
print(f"Fast: {time.time() - start:.4f}s")
```
